```python
df['Sales_Value'] = df['Units_Sold'] * df['Unit_Price']
```

```python
product_sales = df.groupby('SKU_ID')['Sales_Value'].sum().reset_index()
```

```python
product_sales = product_sales.sort_values(
    by='Sales_Value',
    ascending=False
)
```

```python
product_sales.head(10)
```

```python
product_sales.head(20)
```

```python
product_sales['Cumulative_Percentage'] = (
    product_sales['Sales_Value'].cumsum()
    / product_sales['Sales_Value'].sum()
) * 100
```

```python
def abc_category(x):
    if x <= 80:
        return 'A'
    elif x <= 95:
        return 'B'
    else:
        return 'C'

product_sales['ABC_Category'] = (
    product_sales['Cumulative_Percentage']
    .apply(abc_category)
)
```

```python
product_sales.head(20)
```

```python
abc_count = product_sales['ABC_Category'].value_counts()

plt.figure(figsize=(7,5))
plt.bar(abc_count.index, abc_count.values)

plt.title('ABC Analysis of Products')
plt.xlabel('ABC Category')
plt.ylabel('Number of Products')

plt.show()
```

```python
# Show all columns containing inventory or stock information
inventory_cols = [
    col for col in df.columns
    if 'inventory' in col.lower() or 'stock' in col.lower()
]

print(inventory_cols)
```

```python
# Find demand-related columns
demand_cols = [
    col for col in df.columns
    if any(word in col.lower() for word in ['demand', 'sales', 'sold', 'quantity'])
]

# Find inventory-related columns
inventory_cols = [
    col for col in df.columns
    if any(word in col.lower() for word in ['inventory', 'stock'])
]

print("Demand columns:", demand_cols)
print("Inventory columns:", inventory_cols)
```

```python
demand_col = demand_cols[0]
inventory_col = inventory_cols[0]

plt.figure(figsize=(10,5))

plt.scatter(df[demand_col], df[inventory_col])

plt.title('Demand vs Inventory')
plt.xlabel('Demand')
plt.ylabel('Inventory Level')

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df[demand_cols[0]], bins=30)

plt.title('Demand Distribution')
plt.xlabel('Demand')
plt.ylabel('Number of Records')

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.scatter(df[demand_cols[0]], df[inventory_cols[0]])

plt.title('Demand vs Inventory')
plt.xlabel('Demand')
plt.ylabel('Inventory')

plt.show()
```

```python
numeric_cols = df.select_dtypes(include='number').columns

corr = df[numeric_cols].corr()

plt.figure(figsize=(10,7))
plt.imshow(corr, aspect='auto', vmin=-1, vmax=1)

plt.colorbar(label='Correlation')

plt.xticks(
    range(len(numeric_cols)),
    numeric_cols,
    rotation=90
)

plt.yticks(
    range(len(numeric_cols)),
    numeric_cols
)

plt.title('Correlation Matrix')

plt.tight_layout()
plt.show()
```
